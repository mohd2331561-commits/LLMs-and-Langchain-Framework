Q1. What are Large Language Models (LLMs) and how do they function?
Q1 — Large Language Models

Large Language Models (LLMs) are advanced Artificial Intelligence systems designed to understand, process, and generate human language. They are trained on very large collections of text such as books, articles, websites, documents, and other textual data. Examples of LLMs include GPT, Llama, Claude, and Gemini.

Most modern LLMs are based on the Transformer architecture. Transformers use an attention mechanism that helps the model identify relationships between different words or tokens in a sentence or larger piece of text. This allows the model to consider the context of the information while generating a response.

The training process of an LLM involves exposing the model to a large amount of text and allowing it to learn patterns in language. A common training objective is next-token prediction. The model receives a sequence of tokens and learns to predict which token is most likely to appear next. By repeating this process over enormous datasets, the model develops the ability to generate coherent and contextually relevant text.

Before processing text, an LLM divides the input into smaller units called tokens. These tokens are converted into numerical representations that can be processed by the neural network. The model then calculates probabilities for possible next tokens and generates a response one token at a time.

LLMs can perform many tasks, including answering questions, summarizing documents, translating languages, generating content, writing computer programs, explaining concepts, and conducting conversations.

LLMs are useful because they allow people to interact with computer systems using natural language. Instead of learning complicated commands, a user can provide instructions in ordinary language.

However, LLMs do not guarantee that every response will be correct. They may sometimes generate incorrect or fabricated information, known as hallucinations. Therefore, important information should be verified, especially in sensitive areas such as healthcare, law, and finance.

In conclusion, LLMs are powerful AI models that use large-scale training and Transformer-based neural networks to understand patterns in language and generate human-like responses. They have become an important foundation for modern AI applications.

Q2. Discuss the impact of LLMs on traditional software development approaches.
Q2 — Impact of LLMs on Software Development

Large Language Models have significantly changed traditional software development approaches. In conventional software development, programmers manually write source code, search technical documentation, identify errors, create test cases, and prepare documentation. LLMs can now assist developers in many of these activities.

One major impact of LLMs is automatic code generation. Developers can describe a programming requirement using natural language, and an LLM can generate an initial implementation. This is particularly useful for repetitive code, standard functions, database queries, and common programming tasks.

LLMs are also useful for debugging. A developer can provide an error message, stack trace, or problematic section of code and ask the model to identify possible causes and suggest solutions. The model can also explain programming errors in simpler language, which can be useful for students and developers learning new technologies.

Another important application is code explanation. Developers working with an unfamiliar programming language or an existing codebase can use an LLM to understand functions, classes, algorithms, and program structures. This can reduce the time required to understand legacy software.

LLMs also support software testing. They can generate unit tests, test cases, edge cases, and sample inputs. This can help developers identify problems earlier in the software development process.

Documentation is another area that has been improved. LLMs can generate comments, README files, API documentation, technical explanations, and summaries of source code.

LLMs can also assist with database queries, regular expressions, configuration files, programming tutorials, and development planning. As a result, developers can spend more time on system design, architecture, testing, security, and problem solving.

However, AI-generated code has limitations. It can contain bugs, security vulnerabilities, inefficient algorithms, or incorrect assumptions. Therefore, developers must review, test, and validate generated code rather than accepting it without verification.

Overall, LLMs are changing the role of software developers. Instead of replacing programmers completely, they act as intelligent development assistants that can increase productivity and reduce repetitive work. Human developers remain responsible for requirements, architecture, security, quality control, and final decisions.

Q3. What are the key advantages and limitations of using LLMs in real-world applications?
Q3 — Advantages and Limitations of LLMs

Large Language Models provide many advantages in real-world applications, but they also have important limitations. Organizations must understand both aspects before deploying LLM-based systems.

Advantages
Natural Language Interaction: LLMs allow users to communicate with software using normal human language instead of complicated commands.
Automation: LLMs can automate repetitive language-based activities such as summarizing documents, generating reports, drafting emails, and answering common questions.
Increased Productivity: Employees can use LLMs to complete many tasks faster, allowing them to concentrate on more complex activities.
Information Processing: LLMs can process and summarize large amounts of textual information when appropriate context is provided.
Multilingual Capabilities: Many LLMs can understand and generate multiple languages, making them useful for international applications.
Personalization: LLM applications can generate responses according to the user's requirements and the context of the interaction.
Programming Assistance: LLMs can generate code, explain programming concepts, identify errors, and assist with software development.
Limitations
Hallucinations: LLMs can sometimes generate information that sounds convincing but is incorrect or unsupported.
Bias: Models can reproduce biases that exist in their training data.
Privacy Concerns: Organizations must carefully handle confidential and personal information when using external AI services.
Security Risks: LLM applications can face risks such as prompt injection, unauthorized information access, and unsafe tool usage.
Computational Requirements: Large models can require significant computing resources and infrastructure.
Lack of Guaranteed Accuracy: AI-generated information may require human verification, particularly in high-risk fields.
Dependence on Context: The quality of a response depends heavily on the quality of the prompt and the information provided to the model.

Therefore, LLMs can provide substantial benefits in productivity, automation, and information access, but they should be deployed with appropriate security, privacy controls, validation, and human oversight.

Q5. Compare and contrast LangChain and LlamaIndex. What unique problems does each solve?
Q5 — LangChain vs LlamaIndex

LangChain and LlamaIndex are frameworks used to build applications powered by Large Language Models. They can work together, but they have different primary areas of focus.

LangChain mainly focuses on building LLM-powered applications and workflows. It provides components for prompts, language models, chains, agents, tools, memory, and integrations. It is useful when an application requires multiple steps or interactions between an LLM and external tools.

LlamaIndex focuses strongly on connecting LLMs with external data. It provides tools for loading, indexing, storing, and retrieving information from documents and other data sources. It is particularly useful for Retrieval-Augmented Generation (RAG) applications.

Feature	LangChain	LlamaIndex
Primary focus	LLM application workflows	Data indexing and retrieval
Prompt management	Strong	Supported
Chains	Major feature	Less central
Agents	Strong support	Supported
Tool integration	Strong	Supported
Document indexing	Supported	Major strength
RAG	Supported	Major strength
External data	Supported	Core focus
Best suited for	Complex LLM workflows	Document/data-based applications
Problem Solved by LangChain

LangChain helps developers connect multiple LLM-related operations into a structured workflow. For example, an application can receive a question, call an external API, process the result, send information to an LLM, and generate a final response.

Problem Solved by LlamaIndex

LlamaIndex helps developers connect LLMs to external knowledge sources. It can load documents, create indexes, retrieve relevant information, and provide that information as context for an LLM.

Using Both Frameworks

The two frameworks can complement each other. LlamaIndex can handle document loading, indexing, and retrieval, while LangChain can manage the application workflow, prompts, tools, and final LLM response.

Therefore, LangChain is particularly useful for LLM application orchestration, whereas LlamaIndex is particularly useful for connecting LLMs with external data and building retrieval systems.

```python
import os
import requests
from getpass import getpass

os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API key: ")

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


def get_weather(city):

    coordinates = {
        "Mumbai": (19.0760, 72.8777),
        "Delhi": (28.6139, 77.2090),
        "Pune": (18.5204, 73.8567),
        "Bangalore": (12.9716, 77.5946)
    }

    if city not in coordinates:
        return "City not available."

    latitude, longitude = coordinates[city]

    url = (
        f"https://api.open-meteo.com/v1/forecast"
        f"?latitude={latitude}"
        f"&longitude={longitude}"
        f"&current=temperature_2m,"
        f"relative_humidity_2m,wind_speed_10m"
    )

    response = requests.get(url)
    data = response.json()

    current = data["current"]

    return {
        "city": city,
        "temperature": current["temperature_2m"],
        "humidity": current["relative_humidity_2m"],
        "wind_speed": current["wind_speed_10m"]
    }


city = input("Enter city (Mumbai/Delhi/Pune/Bangalore): ")

weather_data = get_weather(city)

print("\nWeather API Data:")
print(weather_data)


llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

prompt = ChatPromptTemplate.from_template(
    """You are a weather assistant.

Use the following weather information retrieved
from a third-party API.

Weather information:
{weather}

Explain the current weather in simple language.
"""
)

chain = prompt | llm | StrOutputParser()

answer = chain.invoke({
    "weather": weather_data
})

print("\nLLM Response:")
print(answer)
```

```python
import os
from getpass import getpass

os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API key: ")


# LlamaIndex
from llama_index.core import (
    VectorStoreIndex,
    SimpleDirectoryReader,
    Settings
)

from llama_index.llms.openai import OpenAI


# LangChain
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


document = """
Python is a high-level programming language known for
its simple syntax and readability. Python is widely used
in web development, data science, artificial intelligence,
automation, and software development.

LangChain is a framework for building applications
powered by language models. It provides components
for prompts, models, chains, agents, and tools.

LlamaIndex is a framework designed to connect language
models with external data. It provides tools for loading,
indexing, and retrieving information from documents.

Retrieval-Augmented Generation, also called RAG, retrieves
relevant information from an external knowledge source
and provides it to a language model as context.
"""

with open("knowledge.txt", "w") as file:
    file.write(document)


# LlamaIndex configuration
Settings.llm = OpenAI(
    model="gpt-4o-mini",
    temperature=0
)

documents = SimpleDirectoryReader(
    input_files=["knowledge.txt"]
).load_data()

index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine()


# LangChain configuration
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

prompt = ChatPromptTemplate.from_template(
    """Answer the user's question using the
retrieved information.

Retrieved information:
{context}

User question:
{question}

Give a clear and concise answer.
"""
)

chain = prompt | llm | StrOutputParser()


# Chatbot
print("Document Q&A Chatbot")
print("Type 'exit' to stop.")

while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        print("Chatbot stopped.")
        break

    retrieved_information = query_engine.query(question)

    answer = chain.invoke({
        "context": str(retrieved_information),
        "question": question
    })

    print("Chatbot:", answer)
```

```python
import os
from getpass import getpass

os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API key: ")


# LlamaIndex imports
from llama_index.core import (
    VectorStoreIndex,
    SimpleDirectoryReader,
    Settings
)

from llama_index.llms.openai import OpenAI


# LangChain imports
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# Create a sample legal document
legal_document = """
LEGAL SERVICE AGREEMENT

This agreement is between ABC Technologies and XYZ Services.

PAYMENT:
The client shall pay the service provider a monthly fee
according to the agreed pricing schedule.

TERMINATION:
Either party may terminate this agreement by providing
30 days written notice to the other party.

CONFIDENTIALITY:
Both parties must keep confidential information received
during the agreement private and must not disclose it to
unauthorized third parties.

LIABILITY:
Each party shall be responsible for losses resulting
from its own negligence or violation of this agreement.

DISPUTE RESOLUTION:
Any dispute should first be resolved through negotiation
between the parties. If negotiation fails, the parties
may use appropriate legal proceedings.
"""

with open("legal_agreement.txt", "w") as file:
    file.write(legal_document)


# Configure LlamaIndex
Settings.llm = OpenAI(
    model="gpt-4o-mini",
    temperature=0
)


# Load the legal document
documents = SimpleDirectoryReader(
    input_files=["legal_agreement.txt"]
).load_data()


# Create document index
index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine()


# Configure LangChain
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)


prompt = ChatPromptTemplate.from_template(
    """You are an AI assistant helping a legal professional.

Use only the retrieved legal document information
to answer the question.

Retrieved information:
{context}

Question:
{question}

Provide a clear and concise answer.
Mention that the original legal document should
be reviewed for final verification.
"""
)


chain = prompt | llm | StrOutputParser()


# Ask questions
questions = [
    "What is the termination period?",
    "What are the confidentiality requirements?",
    "How are disputes resolved?"
]


for question in questions:

    retrieved_information = query_engine.query(question)

    answer = chain.invoke({
        "context": str(retrieved_information),
        "question": question
    })

    print("\n" + "=" * 60)
    print("Question:", question)
    print("Answer:", answer)
```
